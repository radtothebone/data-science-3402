```python
df = pd.read_csv('springleaf-marketing-response/train.csv', low_memory=False)

#df = df.sample(10_000)
```

```python
rows, cols = df.shape
features = [c for c in df.columns if c not in ('ID', 'target')]
num_cols = df[features].select_dtypes(include=[np.number]).columns.tolist()
str_cols = [c for c in features if c not in num_cols]

print('Row and Feature count:')
print(f'Rows: {rows}')
print(f'Columns: {cols}')
print(f'Features: {len(features)}')
```

```python
def group_size(n):
    if n <= 1:
        return 'constant (useless)'
    if n <= 10:
        return 'few values (2-10)'
    if n <= 20:
        return 'relatively diverse values (11-20)'
    return 'many values (21+)'

n_unique = df[features].nunique()
col_type = df[features].dtypes.astype(str).apply(lambda d: 'number' if d.startswith(('int', 'float')) else 'text')
assign_size = n_unique.apply(group_size)
# Unfortunately this ignores date values, we handle dates later.

col_table = pd.DataFrame({'type': col_type, 'distinct values': assign_size})
val_count_groupby = col_table.groupby(['type', 'distinct values']).size()

print('Feature Table')
print(val_count_groupby)
```

```python
# Converting True and False values to numeric, some cols of Bool type will become constant val columns and will be caught by the drop later.
print(f'An example of a constant valued boolean column: \n {df['VAR_0008'].value_counts()}\n')
df = df.replace({True: 1,False: 0})
# After replacement
print(f'Values after replacement: \n {df['VAR_0008'].value_counts()}\n')
```

```python
# useless columns containing no real information.
constant_features = n_unique[n_unique <= 1].index
df_clean = df.drop(columns=constant_features)
var_0008 = True

if 'VAR_0008' not in list(df_clean.columns):
    var_0008 = False

print(f'Dropped {len(constant_features)} constant columns.')
print(f'Is VAR_0008 still in the dataframe? {var_0008}.')
```

```python
# drop_duplicates will destroy original dtypes.
saved_dtypes = df_clean.dtypes

# Transposing the dataframe to take advantage of the drop_duplicates method.
# Would be nice if drop_duplicates had an axis option.
t_pose = df_clean.T
t_pose = t_pose.drop_duplicates()
df_clean = t_pose.T

# Putting the dtypes back, we need the category types to stay.
df_clean = df_clean.astype(saved_dtypes[df_clean.columns])
print(f'Dropped {len(saved_dtypes) - len(df_clean.columns)} duplicate features.')
```

```python
# Dropping columns that are at least almost entirely null, they won't contribute to model performance.
# 250 was chosen arbitrarily.
non_null_min = 250
non_null_df = df_clean.notna().sum()
high_null_cols = non_null_df[non_null_df < non_null_min].index
df_clean = df_clean.drop(columns=high_null_cols)
print(f'Dropped {len(high_null_cols)} columns with fewer than {non_null_min} non-null rows.')
# This dropped less columns than expected.
```

```python
# Working with dates, if a text column looks like a date then we drop it.
# Every date value we saw looks something like 04SEP12:00:00:00, just check for the 3 letter month string.
months = ['JAN', 'FEB', 'MAR', 'APR', 'MAY', 'JUN', 'JUL', 'AUG', 'SEP', 'OCT', 'NOV', 'DEC']

# Non-null vals we look at per col. If even one of them is a date, the whole column gets sent to the shadow realm.
non_null_rows_check = 5
date_cols = []

for col in df_clean.select_dtypes(exclude=[np.number]).columns: # ignore number columns
    sample_values = df_clean[col].dropna().head(non_null_rows_check)
    for value in sample_values:
        if str(value)[2:5] in months:
            # Found a date? You're a date column and we don't want you.
            date_cols.append(col)
            break

# Downside: A column with other data filling up the the data, with a date row further down, will get missed.

# An example of a date column we are dropping.
print(f'Example date column: {date_cols[0]}:')
print(df_clean[date_cols[0]].head())

df_clean = df_clean.drop(columns=date_cols)

print(f'Dropped {len(date_cols)} date columns.')
```

```python
text_cat_cols = []
why_not_one_hot = 0
for col in df_clean.select_dtypes(exclude=[np.number]).columns:
    why_not_one_hot += df_clean[col].nunique()
    df_clean[col] = df_clean[col].astype('category').cat.codes # One of the most important parts of the whole notebook.
    text_cat_cols.append(col)                                  # This converts text variables to numerical so the model can actually learn.

print(f'Columns before cleaning: {df.shape[1]}')
print(f'Columns after cleaning: {df_clean.shape[1]}')
print('One-hot encoding on this dataset would only end in tears. ')
print(f'The {len(text_cat_cols)} text columns would have made {why_not_one_hot} new columns. ')
```

```python
print('For our choice of ML model, rescaling is not needed. HistGradientBoostingClassifier works well')
print('with unscaled data. ')
```

```python
# Saving a copy of one column so we can see the before and after.
example_col = 'VAR_0602'
before_values = df_clean[example_col].copy()

numeric_col = df_clean.select_dtypes(include=[np.number]).columns.drop(['ID', 'target'])

replaced_cells = 0

# We will naievely convert the assumed placeholder values to na.
for col in numeric_col:
    for placeholder in assumed_placeholders:
        count = (df_clean[col] == placeholder).sum()
        if count > 0:
            df_clean[col] = df_clean[col].replace(placeholder, np.nan)
            replaced_cells += count

print('Placeholder handling:')
print('Each column is checked against the placeholder list one value at a time. ')
print(f'Replaced {replaced_cells} placeholder values with NaN. ')

plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
before_values.plot.hist(bins=50)
plt.title(f'{example_col} before')

plt.subplot(1, 2, 2)
df_clean[example_col].plot.hist(bins=50)
plt.title(f'{example_col} after')

plt.show()
```

```python
# Separating the target and ID columns from the model features.
X = df_clean.drop(columns=['ID', 'target'])
y = df_clean['target']

# We used a more "manual" version of this in class, but I prefer this method.
# Splitting 60% of the data into training data and leaving 40% for validation and testing.
X_train, X_rest, y_train, y_rest = train_test_split(X, y, test_size=0.4, random_state=0, stratify=y)

# Splitting the remaining 40% evenly between validation and testing data.
X_val, X_test, y_val, y_test = train_test_split(X_rest, y_rest, test_size=0.5, random_state=0, stratify=y_rest)

# The test.csv from Kaggle isn't useful for us here since it doesn't have the target variable, so we just have to split our split data.

# Printing the number of rows in each dataset.
print('Splitting into train, validation, and test (60/20/20):')
print(f'train: {len(X_train)} rows. ')
print(f'validation: {len(X_val)} rows. ')
print(f'test: {len(X_test)} rows. ')

print('Evaluation on the validation set:')
print('Kaggle scored this problem on AUC, which is why we selected it here. \n')
```

```python
# Parameters chosen as a starting point, no real logic here. Just trying to get a non-trivial result.
model = HistGradientBoostingClassifier(max_iter=500, learning_rate=0.05, early_stopping=True, validation_fraction=0.1, random_state=0)
model.fit(X_train, y_train)

print('HistGradientBoostingClassifier was chosen for four reasons:')
print('I (Christian) happened to randomly choose it for lab 8, so I\'m a little familiar with it. ')
print('It handles NaN without issue, so missing data doesn\'t need imputing, no rescaling, since ')
print('its tree-based, and apparently its well-suited for very wide data. ')
print(f'Training stopped after {model.n_iter_} of {model.max_iter} iterations. ')

print('Question: Is the model overfitting?')

# Prediction probabilities for each set.
train_scores = model.predict_proba(X_train)[:, 1]
val_scores = model.predict_proba(X_val)[:, 1]
test_scores = model.predict_proba(X_test)[:, 1]

# AUC scores.
train_auc = roc_auc_score(y_train, train_scores)
val_auc = roc_auc_score(y_val, val_scores)
test_auc = roc_auc_score(y_test, test_scores)

# We won't use accuracy because about 77% of the target is class 0.
# If the model always guessed 0, it would get a C. We don't want that.
dont_use_accuracy = accuracy_score(y_test, model.predict(X_test))

# The points used to draw each ROC curve.
train_fpr, train_tpr, thresholds = roc_curve(y_train, train_scores)
val_fpr, val_tpr, thresholds = roc_curve(y_val, val_scores)
test_fpr, test_tpr, thresholds = roc_curve(y_test, test_scores)

# One curve per dataset, plus a line for random guessing to compare against.
plt.plot(train_fpr, train_tpr, label=f'Training (AUC = {train_auc:.4f})')
plt.plot(val_fpr, val_tpr, label=f'Validation (AUC = {val_auc:.4f})')
plt.plot(test_fpr, test_tpr, label=f'Test (AUC = {test_auc:.4f})')
plt.plot([0, 1], [0, 1], linestyle='dotted', color='red', label='random guess')

plt.xlabel('FPR')
plt.ylabel('TPR')
plt.title('ROC Curves by Dataset')
plt.legend()
plt.show()

print(f'Training AUC: {train_auc:.4f}')
print(f'Validation AUC: {val_auc:.4f}')
print(f'Test AUC: {test_auc:.4f}')
print(f'Test accuracy (dont_use_accuracy): {dont_use_accuracy:.4f}\n')

print('Our earlier conclusion was wrong in the presentation, the model is overfitting.') # We looked at the wrong metrics.
print('The gap between the training AUC and the test AUC is noticeable.')
```

```python
print('Trying to correct the overfitting:')
model_corrected = HistGradientBoostingClassifier(max_iter=500, max_leaf_nodes=7, min_samples_leaf=200, n_iter_no_change=15, learning_rate=0.05, early_stopping=True, validation_fraction=0.1, random_state=0)
model_corrected.fit(X_train, y_train)
# We did not optimize these new parameters fully. Just tried something new that would pull back on the overfitting.
# We made the model a little more simple by reducing the tree complexity, requiring "bigger" leaves, and stopping training when earlier when it stopps improving.

# Prediction probabilities for each dataset.
train_scores_corrected = model_corrected.predict_proba(X_train)[:, 1]
val_scores_corrected = model_corrected.predict_proba(X_val)[:, 1]
test_scores_corrected = model_corrected.predict_proba(X_test)[:, 1]

# AUC scores.
train_auc_corrected = roc_auc_score(y_train, train_scores_corrected)
val_auc_corrected = roc_auc_score(y_val, val_scores_corrected)
test_auc_corrected = roc_auc_score(y_test, test_scores_corrected)

# The points used to draw each ROC curve.
train_fpr_corrected, train_tpr_corrected, thresholds_corrected = roc_curve(y_train, train_scores_corrected)
val_fpr_corrected, val_tpr_corrected, thresholds_corrected = roc_curve(y_val, val_scores_corrected)
test_fpr_corrected, test_tpr_corrected, thresholds_corrected = roc_curve(y_test, test_scores_corrected)

# One curve per dataset, plus a line for random guessing to compare against.
plt.plot(train_fpr_corrected, train_tpr_corrected, label=f'Training (AUC = {train_auc_corrected:.4f})')
plt.plot(val_fpr_corrected, val_tpr_corrected, label=f'Validation (AUC = {val_auc_corrected:.4f})')
plt.plot(test_fpr_corrected, test_tpr_corrected, label=f'Test (AUC = {test_auc_corrected:.4f})')
plt.plot([0, 1], [0, 1], linestyle='dotted', color='red', label='random guess')

plt.xlabel('FPR')
plt.ylabel('TPR')
plt.title('ROC Curves by Dataset')
plt.legend()
plt.show()

print(f'Training AUC: {train_auc_corrected:.4f}')
print(f'Validation AUC: {val_auc_corrected:.4f}')
print(f'Test AUC: {test_auc_corrected:.4f}')

print('\n The overfitting correction was generally successful. Doing some research, there were three')
print('model parameters that seemed suited for tuning the model.')
print('max_leaf_nodes, min_samples_leaf, n_iter_no_change were the top parameters for pulling back on the model.')
```

```python
print('Question: How does the non-overfitting model do on each target?')

test_pred = model_corrected.predict(X_test)

print('Confusion matrix on the test set')
cm = confusion_matrix(y_test, test_pred)

print('The model correctly predicts 1895 True Positives, and 21187 true negatives.')
cm_df = pd.DataFrame(cm, columns=['Predicted Negative', 'Predicted Positive'], index=['Actual Negative', 'Actual Positive'])
display(cm_df)
```

```python
print('Question: How much did the overfitting improve?')

model_comparison = pd.DataFrame({
    'Training AUC': [train_auc, train_auc_corrected],
    'Test AUC': [test_auc, test_auc_corrected],
    'Train - Test Gap': [train_auc - test_auc, train_auc_corrected - test_auc_corrected],
}, index=['Original model', 'Improved model'])

display(model_comparison.round(4))

print('The improved model gives up nearly no test AUC, but the huge gap between training and test is no more.')
```
